# PlantCV color, scale and crop
Run from the repository root. Set paths in the cells and set RUN to True.

```python
from pathlib import Path
import subprocess
import sys

USE_DRIVE = False
if USE_DRIVE:
    from google.colab import drive
    drive.mount("/content/drive")
    DATA = Path("/content/drive/MyDrive/research")
else:
    DATA = Path("private_data")
RUN = False
```

```python
if RUN:
    subprocess.run([sys.executable, "01_preprocessing/plantcv_preprocessing.py",
                    "normalize", "--input", str(DATA / "raw_images"),
                    "--output", str(DATA / "normalized_images")], check=True)
```

```python
CROP_INPUT = DATA / "normalized_images"
if RUN:
    subprocess.run([sys.executable, "01_preprocessing/plantcv_preprocessing.py",
                    "crop", "--input", str(CROP_INPUT), "--output", str(DATA / "cropped_images"),
                    "--crop-mode", "leaf_on_white"], check=True)
```
